```python
output_dir = Path('/home/yuan/omics/output')
output_pdb_dir = output_dir / 'pdb'
simulate_dir = output_dir / 'simulate'

combo = 'combo5'
indir = simulate_dir / f'{combo}_relaxed'
for root, dirs, files in os.walk(indir):
    for file_name in files:
        if file_name == 'fixed.pdb':
            outdir = Path(root.replace(f'simulate/{combo}_relaxed', f'pdb/{combo}_fixed'))
            outdir.mkdir(parents=True, exist_ok=True)
            # pdb
            infile = Path(root) / file_name
            outfile = outdir / file_name
            infile.replace(outfile)

            json_name = 'fixed_info.json'
            infile = Path(root) / json_name
            if infile.is_file():
                outfile = outdir / json_name
                infile.replace(outfile)    
```
