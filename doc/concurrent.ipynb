```python
import asyncio


async def f1(data):
    print(f"f1 started: {data}")
    await asyncio.sleep(3)   # simulate work
    print(f"f1 finished: {data}")
    return f"f1 result: {data}"


async def f2(data):
    print(f"f2 started: {data}")
    await asyncio.sleep(1)   # simulate work
    print(f"f2 finished: {data}")
    return f"f2 result: {data}"


async def worker(name, func, queue):

    while True:
        data = await queue.get()

        if data is None:
            queue.task_done()
            break

        try:
            result = await func(data)
            print(result)

        finally:
            queue.task_done()


async def main():

    data = ["A", "B", "C", "D", "E", "F", "G", "H"]

    queue = asyncio.Queue()

    # Put input data into queue
    for item in data:
        await queue.put(item)

    # Two workers
    task1 = asyncio.create_task(
        worker("worker1", f1, queue)
    )

    task2 = asyncio.create_task(
        worker("worker2", f2, queue)
    )

    # Wait until all data has been processed
    await queue.join()

    # Stop workers
    await queue.put(None)
    await queue.put(None)

    await asyncio.gather(task1, task2)


asyncio.run(main())
```
